Создайте класс StepCounter, который хранит количество шагов пользователя.

Используйте приватный атрибут _steps

Добавьте методы:

- add_steps(amount) — добавляет шаги (не может быть отрицательным) В случае ошибки выдается сообщение: "Ошибка! Количество шагов не может быть отрицательным!"

- reset() — сбрасывает счётчик до 0

- get_steps() — возвращает текущее значение

````python
class StepCounter:

    def __init__(self, _steps):
      if _steps >= 0:
          self._steps = _steps
      else:
          self._steps = 0

    def add_steps(self, amount):
      if amount < 0:
          print("Ошибка! Количество шагов не может быть отрицательным!")
      else:
          self._steps += amount

    def reset(self):
      self._steps = 0

    def get_steps(self):
      return self._steps

count = StepCounter(88)
count.add_steps(12)
print(count.get_steps())
````

Создайте класс Counter с атрибутом value, который по умолчанию равен 0.

Добавьте следующие методы:

- increment() — увеличивает значение на 1;

- decrement() — уменьшает значение на 1, но не меньше 0;

- reset() — обнуляет счётчик.

Пример работы:

```
c = Counter()
c.increment()
c.increment()
c.decrement()
print(c.value)  # 1
c.reset()
print(c.value)  # 0
```



````python
class Counter:
    def __init__(self):
      self.value = 0

    def increment(self):
      self.value += 1

    def decrement(self):
      if self.value > 0:
        self.value -= 1

    def reset(self):
      self.value = 0

c = Counter()
c.increment()
c.increment()
c.decrement()
print(c.value)  # 1
c.reset()
print(c.value)  # 0
````

Класс Employee и Manager

Employee имеет атрибуты: имя, должность, зарплата

- Метод: get_info(), возвращает информацию об сотруднике

Manager наследуется от Employee, добавляет атрибут team (список сотрудников)

- Метод: add_member(employee) — добавить сотрудника в команду

Переопределить get_info() так, чтобы показывать количество подчинённых

````python
class Employee:
  def __init__(self, name, position, salary):
    self.name = name
    self.position = position
    self.salary = salary

  def get_info(self):
    return (f"Сотрудник: {self.name}, Должность: {self.position}, Зарплата: {self.salary}")

class Manager(Employee):
  def __init__(self, name, position, salary):
    super().__init__(name, position, salary)
    self.team = []

  def add_member(self, employee):
    self.team.append(employee)

  def get_info(self):
    base_info = super().get_info()
    return f"{base_info}, Подчиненных: {len(self.team)}"

worker1 = Employee("Иван", "Учитель", 80000)
print(worker1.get_info())
boss = Manager("Анна", "Директор", 100000)
boss.add_member(worker1)
print(boss.get_info())
````

## **Задача 2**

Класс BankAccount

Атрибуты: держатель (имя), баланс

Методы:
  - deposit(amount),
  
  - withdraw(amount)

Проверка: нельзя снять больше, чем есть на счёте

Метод: __str__ для красивого вывода (например, "BankAccount(Иван, баланс=1000)")

````python
class BankAccount:
  def __init__(self, holder, balance):
    self.holder = holder
    self.balance = balance

  def deposit(self, amount):
    if amount > 0:
      self.balance += amount
    return self.balance

  def withdraw(self, amount):
    if 0 < amount <= self.balance:
      self.balance -= amount
    else:
      print(f"Ошибка: недостаточно средств. На счете только {self.balance}.")
    return self.balance

  def __str__(self):
    return f"BankAccount({self.holder}, баланс={self.balance})"

  account = BankAccount("Иван", 1000)
  print(account)
````

## **Задача 4**

Создайте два класса:

**Класс Person**

Описание:

Базовый класс с атрибутами:

- name — имя человека,

- age — возраст.

Для обоих атрибутов используйте @property и сеттеры с проверкой:

- имя — непустая строка;

- возраст — положительное целое число.

Реализуйте метод __repr__, который возвращает строку:

Person(name='Анна', age=25)


**Класс Student(Person)**

Описание:

Наследуется от Person.

Добавляет:

- атрибут grades — список оценок (по умолчанию пустой список);

- метод add_grade(grade) — добавляет оценку (только числа от 0 до 5);

- свойство average (только для чтения, через @property), которое возвращает средний балл (или 0, если оценок нет).

Переопределите __repr__, чтобы выводилось имя, возраст и средний балл:

Student(name='Анна', age=20, average=4.5)

Пример использования:

```
p = Person("Иван", 30)
print(p)
# ➜ Person(name='Иван', age=30)

s = Student("Анна", 19)
s.add_grade(5)
s.add_grade(4)
print(s)
# ➜ Student(name='Анна', age=19, average=4.5)
```

````python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  @property
  def name(self):
    return self._name

  @name.setter
  def name(self, value):
    if not isinstance(value, str) or not value.strip():
        raise ValueError("Имя должно быть непустой строкой.")
    self._name = value

  @property
  def age(self):
    return self._age

  @age.setter
  def age(self, value):
    if not isinstance(value, int) or value <= 0:
      raise ValueError("Возраст должен быть положительным целым числом.")
      self._age = value

  def __repr__(self):
    return f"Person(name='{self.name}', age={self.age})"

class Student(Person):
  def __init__(self, name, age):
    super().__init__(name, age)
    self._grades = []

  def add_grade(self, grade):
    if not isinstance(grade, (int, float)) or not (0 <= grade <= 5):
      raise ValueError("Оценка должна быть числом от 0 до 5.")
    self._grades.append(grade)

  @property
  def average(self):
    if not self._grades:
      return 0
    return sum(self._grades) / len(self._grades)

  def __repr__(self):
    avg_formatted = round(self.average, 1)
    return f"Student(name='{self.name}', age={self.age}, average={avg_formatted})"



p = Person("Иван", 30)
print(p)

s = Student("Анна", 19)
s.add_grade(5)
s.add_grade(4)
print(s)
````

Создайте класс SchoolJournal с атрибутами students — список студентов.

Добавьте методы:

- add_student(student) — добавить ученика;

- get_best_student() — вернуть ученика с наивысшим средним баллом;

- show_all() — вывести всех студентов и их средний балл.

````python
class Student:
  def __init__(self, name, grades=None):
    self.name = name
    self.grades = grades if grades is not None else []

  def add_grade(self, grade):
    self.grades.append(grade)

  def average(self):
    if not self.grades:
      return 0.0
    return round(sum(self.grades) / len(self.grades), 2)


class SchoolJournal:
  def __init__(self):
    self.students = []

  def add_student(self, student):
    self.students.append(student)

  def get_best_student(self):
    if not self.students:
        return None

    best_student = max(self.students, key=lambda s: s.average())
    return best_student

  def show_all(self):
    if not self.students:
        print("Журнал пуст.")
        return
    for student in self.students:
      print(f"Студент: {student.name}, Средний балл: {student.average()}")


s1 = Student("Анна", [5, 4, 5])
s2 = Student("Борис", [3, 4, 4])
s3 = Student("Виктор")

s3.add_grade(5)
s3.add_grade(5)

journal = SchoolJournal()
journal.add_student(s1)
journal.add_student(s2)
journal.add_student(s3)

journal.show_all()
best = journal.get_best_student()
print(f"\nЛучший студент: {best.name} (балл: {best.average()})")
````

Создайте класс Course, представляющий учебный курс, и класс Student, который может записываться на курсы и получать оценки.

**Класс Course**

Атрибуты:

- students — список имён студентов (по умолчанию пустой).

Реализовать:

Course(title='Python', students=2/10)

**Класс Student**

Атрибуты:

- courses — словарь {название_курса: оценка}.

Реализовать:

- add_course(course) — добавляет объект Course в список курсов студента (добавляет его имя на курс);

- rate_course(course_name, grade) — выставляет оценку (0–5);

- свойство average — средняя оценка по всем курсам;

- __repr__ — выводит имя и средний балл.


```
python = Course("Python", 3)
ml = Course("Machine Learning", 2)

s1 = Student("Анна")
s2 = Student("Борис")

s2.add_course(python)

print(python)       # ➜ Course(title='Python', students=2/3)

```

````python
class Course:
  def __init__(self, title, max_students):
    self.title = title
    self.max_students = max_students
    self.students = []

  def add_student(self, name):
    if len(self.students) < self.max_students:
        self.students.append(name)
        return True
    else:
        print(f"Ошибка: на курс '{self.title}' больше нельзя записаться (лимит {self.max_students}).")
        return False

  def __len__(self):
    return len(self.students)

  def __repr__(self):
    return f"Course(title='{self.title}', students={len(self)}/{self.max_students})"


class Student:
  def __init__(self, name):
    self.name = name
    self.courses = {}

  def add_course(self, course):
    if course.add_student(self.name):
      self.courses[course.title] = None
    else:
      print(f"Студент {self.name} не смог записаться на курс '{course.title}'.")

  def rate_course(self, course_name, grade):
    if not (0 <= grade <= 5):
        print("Ошибка: оценка должна быть от 0 до 5.")
        return


    if course_name in self.courses:
        self.courses[course_name] = grade
    else:
        print(f"Ошибка: студент {self.name} не записан на курс '{course_name}'.")

  @property
  def average(self):
    valid_grades = [grade for grade in self.courses.values() if grade is not None]
    if not valid_grades:
      return 0.0

    return round(sum(valid_grades) / len(valid_grades), 2)

  def __repr__(self):
    return f"Student(name='{self.name}', average={self.average})"


python = Course("Python", 3)
ml = Course("Machine Learning", 2)

s1 = Student("Анна")
s2 = Student("Борис")

s1.add_course(python)
s2.add_course(python)
s1.add_course(ml)

s1.rate_course("Python", 5)
s1.rate_course("Machine Learning", 4)

print(s1)
print(len(python))
print(python)
````

````python
class Song:
  def __init__(self, title, artist, duration):
    self.title = title
    self.artist = artist
    self.duration = duration

  @property
  def title(self):
    return self._title

  @title.setter
  def title(self, value):
    if not isinstance(value, str) or not value.strip():
        raise ValueError("Название песни должно быть непустой строкой.")
    self._title = value

  @property
  def artist(self):
    return self._artist

  @artist.setter
  def artist(self, value):
    if not isinstance(value, str) or not value.strip():
      raise ValueError("Имя исполнителя должно быть непустой строкой.")
    self._artist = value

  @property
  def duration(self):
    return self._duration

  @duration.setter
  def duration(self, value):
    if not isinstance(value, (int, float)) or value <= 0:
      raise ValueError("Длительность должна быть положительным числом.")
    self._duration = value

  def __repr__(self):
    return f"Song(title='{self.title}', artist='{self.artist}', duration={self.duration})"

class Playlist:
  def __init__(self, name):
    self.name = name
    self.songs = []

  def add_song(self, song):
    if not isinstance(song, Song):
      raise TypeError("В плейлист можно добавлять только объекты класса Song.")
    self.songs.append(song)

  def __len__(self):
    return len(self.songs)

  def __getitem__(self, index):
    return self.songs[index]

  def __repr__(self):
    return f"Playlist(name='{self.name}', songs={len(self.songs)})"

s1 = Song("Imagine", "John Lennon", 183)
s2 = Song("Yesterday", "The Beatles", 145)
pl = Playlist("Classic Rock")

pl.add_song(s1)
pl.add_song(s2)

print(pl)          # ➜ Playlist(name='Classic Rock', songs=2)
print(len(pl))     # ➜ 2
print(pl[0])       # ➜ Song(title='Imagine', artist='John Lennon', duration=183)
````
